# Cálculo de SVD con el método unilateral de rotaciones de Jacobi

El siguiente código corresponde a una implementación del algoritmo unilateral de rotaciones de Jacobi para calcular SVD de una matriz rectangular.

```python
import numpy as np

"""
- Entrada: Matriz X (N x D)
- Salida: Matrices U, S y V,
          tales que: X = U * S * V^T 
                     S es diagonal
"""
def jacobi_svd(X : int, iterations:int = 100):
    X = np.array(X, dtype=float)
    N, D = X.shape  # N patrones x D dimensiones/características
    U = X.copy()  # Inicialmente, U = X
    V = np.eye(D)  # Inicialmente, V = identidad

    for _ in range(iterations):
        for i in range(D - 1):  # Para cada dimensión i < (D - 1)...
            for j in range(i + 1, D):  # Para cada dimensión j tal que j < D y j > i...
                # normas columnas
                a = np.linalg.norm(U[:, i])
                b = np.linalg.norm(U[:, j])

                if a < b:
                    U[:, [i, j]] = U[:, [j, i]]
                    V[:, [i, j]] = V[:, [j, i]]

                x = np.dot(U[:, i], U[:, i])
                y = np.dot(U[:, j], U[:, j])
                w = np.dot(U[:, i], U[:, j])

                if w != 0:
                    alpha = (y - x) / (2 * w)
                    denom = abs(alpha) + np.sqrt(1 + alpha**2)

                    if alpha >= 0:
                        t = 1.0 / denom
                    else:
                        t = -1.0 / denom

                    c = 1.0 / np.sqrt(1 + t**2)
                    s = c * t
                          
                else:
                    c = 1.0
                    s = 0.0

                Ui = U[:, i].copy()
                Uj = U[:, j].copy()
                U[:, i] = c * Ui - s * Uj
                U[:, j] = s * Ui + c * Uj

                Vi = V[:, i].copy()
                Vj = V[:, j].copy()
                V[:, i] = c * Vi - s * Vj
                V[:, j] = s * Vi + c * Vj

    singvals = np.zeros(D)
    for j in range(D):
        singvals[j] = np.linalg.norm(U[:, j])
        if singvals[j] > 0:
            U[:, j] /= singvals[j]

    S = np.diag(singvals)
    return U, S, V.T
```
