## LLE 2D sobre MNIST

```python
from sklearn.datasets import fetch_openml
from sklearn.manifold import LocallyLinearEmbedding
import matplotlib.pyplot as plt


mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X = (mnist.data / 255.0)[0:5000]
y = (mnist.target.astype(int))[0:5000]

lle = LocallyLinearEmbedding(n_neighbors=15)
lle_res = lle.fit_transform(X)

plt.scatter(lle_res[:, 0], lle_res[:, 1], c=y, cmap='jet', s=1)
plt.title('LLE of MNIST Dataset (2 Components)')
plt.xlabel('LLE 2D Component 1')
plt.ylabel('LLE 2D Component 2')
plt.colorbar(label='Digit')
plt.show()
```

# ISOMAP
## ISOMAP 2D sobre un rollo suizo

```python
from sklearn.datasets import make_swiss_roll
from sklearn.manifold import Isomap
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


X, color = make_swiss_roll(n_samples=1500, noise=0.2)
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=color, cmap=plt.cm.viridis)
ax.set_title("Swiss Roll")
plt.show()


isomap = Isomap(n_neighbors=15, n_components=2)
isomap_res = isomap.fit_transform(X)


plt.scatter(isomap_res[:, 0], isomap_res[:, 1], s=1, c=color)
plt.title('ISOMAP of Swiss Roll')
plt.xlabel('ISOMAP 2D Component 1')
plt.ylabel('ISOMAP 2D Component 2')
plt.show()
```

# t-SNE
## t-SNE 2D sobre MNIST

```python
from sklearn.datasets import fetch_openml
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
```

```python
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X = (mnist.data / 255.0)[0:5000]
y = (mnist.target.astype(int))[0:5000]
```

```python
tsne = TSNE(n_components=2, verbose=1, perplexity=70, max_iter=300)
tsne_results = tsne.fit_transform(X)
```

```python
plt.scatter(tsne_results[:, 0], tsne_results[:, 1], c=y, cmap='jet', s=1)
plt.title('t-SNE of MNIST Dataset (2 Components)')
plt.xlabel('t-SNE 2D Component 1')
plt.ylabel('t-SNE 2D Component 2')
plt.colorbar(label='Digit')
plt.show()
```

# UMAP
## UMAP 2D sobre MNIST

```python
from sklearn.datasets import fetch_openml
from umap import UMAP
import matplotlib.pyplot as plt
```

```python
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X = (mnist.data / 255.0)[0:5000]
y = (mnist.target.astype(int))[0:5000]
```

```python
umap = UMAP(n_neighbors=15, n_components=2, random_state=42)
umap_results = umap.fit_transform(X)
```

```python
plt.scatter(umap_results[:, 0], umap_results[:, 1], c=y, cmap='jet', s=1)
plt.title('UMAP of MNIST Dataset (2 Components)')
plt.xlabel('UMAP 2D Component 1')
plt.ylabel('UMAP 2D Component 2')
plt.colorbar(label='Digit')
plt.show()
```
